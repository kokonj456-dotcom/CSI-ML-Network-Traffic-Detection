# การสำรวจเชิงสถิติ CIC-IDS2017 (Exploratory Data Analysis)

สำรวจการกระจายตัวของระยะเวลาเชื่อมต่อ (Duration), ปริมาณไบต์ (Forward/Backward Bytes), 
โปรโตคอล (Protocol), และพฤติกรรมการโจมตีตามมาตรฐาน MITRE ATT&CK


## 1. โหลดข้อมูล Harmonized สำหรับ EDA


```python
import pandas as pd
import numpy as np
from pathlib import Path

harm_dir = Path("output/harmonized")
features = pd.read_csv(harm_dir / "features.csv")
labels = pd.read_csv(harm_dir / "labels.csv")
meta = pd.read_csv(harm_dir / "metadata.csv")

print(f"Harmonized Records: {len(features):,}")
```

## 2. การกระจายตัวของขนาดทราฟฟิกและระยะเวลา


```python
stats = features[["duration", "bytes_forward", "bytes_backward"]].describe(percentiles=[0.25, 0.5, 0.75, 0.95, 0.99])
display(stats)
```

## 3. สรุปความถี่ของประเภทการโจมตี (Attack Taxonomy)


```python
cat_summary = labels["attack_category"].value_counts().to_frame("count")
cat_summary["percentage"] = (cat_summary["count"] / len(labels) * 100).round(2)
display(cat_summary)
```
